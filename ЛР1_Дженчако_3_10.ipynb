### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]])

mean=np.mean(A, axis=0)
std=np.std(A, axis=0)

print((A-mean)/std)
````
