## Завдання 8. Визначник та обернена матриця

Для матриці

\[
A =
\begin{bmatrix}
4 & 7\\
2 & 6
\end{bmatrix}
\]

1. Знайдіть визначник.
2. Знайдіть обернену матрицю.
3. Перевірте результат множенням `A @ A_inv`.
4. Поясніть у коментарі, який результат має бути отриманий.

````python
A = np.array([[4, 7],
              [2, 6]])

# Ваш код тут
print(np.linalg.det(A))
a_inv=np.linalg.inv(A)
print(a_inv)
print(A@a_inv) #повинна вийти одиинчна матриця
````

## Завдання 13. Функція з умовою

Створіть функцію

\[
f(x) =
\begin{cases}
x^2, & x < 0 \\
2x + 1, & x \geq 0
\end{cases}
\]

Застосуйте її до вектора значень від `-5` до `5`.

Спробуйте реалізувати функцію двома способами:
1. через звичайну функцію Python;
2. через `np.where()`.

````python
x = np.arange(-5, 6)

# Варіант 1
def vector(m):
    for j in range(len(m)):
        if m[j]<0:
            m[j]=m[j]**2
        else:
            m[j]=2*m[j]+1
    return m
k=vector(x)
print(k)
# Варіант 2
x = np.arange(-5, 6)
x2=np.where(x<0, x**2, 2*x+1)
print(x2)
````
